### 1. Tính $L_p$-norm giữa hai điểm $(1, 2)$ và $(3, 4)$ với $p = 1, 2, \infty$.

<!-- Tính $L_p$-norm giữa hai điểm $(1, 2)$ và $(3, 4)$ với $p = 1, 2, \infty$. -->
> Theo giải tích 2A đã học

Xét trong không gian metric $\mathbb{R}^2$, ta định nghĩa $L_p$ là khoảng cách giữa hai điểm $x = (x_1, x_2)$ và $y = (y_1, y_2)$ được tính theo công thức sau:
$$L_p = \|x - y\|_p = \left( \sum_{k=1}^{2} |x_k - y_k|^p \right)^{\frac{1}{p}} $$

- Với p = 1 thì $L_1$ được gọi là khoảng cách Manhattan, được tính bằng tổng giá trị tuyệt đối của hiệu giữa các thành phần của hai điểm:
$$L_1 = |x_1 - y_1| + |x_2 - y_2| $$

- Với p = 2 thì $L_2$ được gọi là khoảng cách Euclid, được tính bằng căn bậc hai của tổng bình phương của hiệu giữa các thành phần của hai điểm:
$$L_2 = \sqrt{(x_1 - y_1)^2 + (x_2 - y_2)^2} $$

- Với p = $\infty$ thì $L_\infty$ được gọi là khoảng cách tối đa, được tính bằng giá trị tuyệt đối lớn nhất của hiệu giữa các thành phần của hai điểm:
$$L_\infty = \max\{|x_1 - y_1|, |x_2 - y_2|\} $$

$$\text{Bài làm}$$
Cho x = (1, 2) và y = (3, 4), ta có:
-  $L_1$:
$$L_1 = |1 - 3| + |2 - 4| = 2 + 2 = 4$$
-  $L_2$:
$$L_2 = \sqrt{(1 - 3)^2 + (2 - 4)^2} = \sqrt{4 + 4} = \sqrt{8} = 2\sqrt{2}$$
-  $L_\infty$:
$$L_\infty = \max\{|1 - 3|, |2 - 4|\} = \max\{2, 2\} = 2$$

### 2. Chứng minh rằng khoảng cách Mahalanobis giữa hai điểm dữ liệu tương đương với khoảng cách Euclidean trên một tập dữ liệu đã được biến đổi, trong đó phép biến đổi được thực hiện bằng cách biểu diễn dữ liệu dọc theo các thành phần chính (principal components) và chia cho độ lệch chuẩn của từng thành phần.

Trước hết khoảng cách Mahalanobis giữa hai điểm dữ liệu $x$ và $y$ được định nghĩa như sau:
$$D_M(x, y) = \sqrt{(x - y)^T \Sigma^{-1} (x - y)}$$
Trong đó $\Sigma$ là ma trận hiệp phương sai của dữ liệu. Mà ma trận hiệp phương sai $\Sigma$ là ma trận đối xứng và khả nghịch nên có thể được phân rã thành ma trận trực giao $P$ chứa các vector riêng (eigenvectors) và ma trận đường chéo $D$ chứa các giá trị riêng (eigenvalues) như sau: 
$$\Sigma = PDP^T$$
Ta nghịch đảo ma trận hiệp phương sai $\Sigma$:
$$\Sigma^{-1} = (PDP^T)^{-1} = P D^{-1} P^T$$
Thay vào công thức khoảng cách Mahalanobis, ta có:
$$D_M(x, y) = \sqrt{(x - y)^T P D^{-1} P^T (x - y)}$$
Ta bình phương khoảng cách Mahalanobis:
$$D_M^2(x, y) = (x - y)^T P D^{-1} P^T (x - y) = (x - y)^T P D^{-1/2} D^{-1/2} P^T (x - y)$$
Đặt x' = $D^{-1/2} P^T x$ và y' = $D^{-1/2} P^T y$, Ta có:
$$x' - y' = D^{-1/2} P^T (x - y)$$
Do $(AB)^T = B^T A^T$ và $D^{-1/2}$ là ma trận đối xứng, ta có: $$(x' - y')^T  = (x - y)^T P D^{-1/2}$$

Thay vào công thức khoảng cách Mahalanobis, ta có:
$$D_M^2(x, y) = (x' - y')^T (x' - y')$$
Mà trong không gian metric $\mathbb{R}^n$, khoảng cách Euclidean giữa hai điểm $x'$ và $y'$ được định nghĩa là:
$$L_2(x', y') = \|x' - y'\|_2 = \sqrt{(x' - y')^T (x' - y')}$$
Do đó, ta có:
$$L_2(x', y')^2 = \|x' - y'\|_2^2 = (x' - y')^T (x' - y') = D_M^2(x, y)$$
Vì vậy:
$$D_M(x, y) = L_2(x', y')$$

### 4. Tính độ tương đồng khớp (match-based similarity), độ tương đồng cosine (cosine similarity), và hệ số Jaccard giữa hai tập hợp $\{A, B, C\}$ và $\{A, C, D, E\}$.

Gọi tập hợp 1 là $S_1 = \{A, B, C\}$ và tập hợp 2 là $S_2 = \{A, C, D, E\}$.

Ta có: $S_1 \cap S_2 = \{A, C\}$ và $S_1 \cup S_2 = \{A, B, C, D, E\}$

> Match-based similarity

Match($S_1, S_2$) = $|S_1 \cap S_2| = 2$

> Cosine similarity

Để tính độ tương đồng cosine, ta cần biểu diễn các tập hợp dưới dạng vector nhị phân. Ta có:
- $S_1$ được biểu diễn thành vector $v_1 = (1, 1, 1, 0, 0)$
- $S_2$ được biểu diễn thành vector $v_2 = (1, 0, 1, 1, 1)$
Độ tương đồng cosine được tính bằng công thức:
$$\text{Cosine}(v_1, v_2) = \frac{v_1 \cdot v_2}{\|v_1\| \|v_2\|}$$
Tính tích vô hướng $v_1 \cdot v_2$:
$$v_1 \cdot v_2 = (1)(1) + (1)(0) + (1)(1) + (0)(1) + (0)(1) = 2$$
Tính chuẩn của $v_1$ và $v_2$:
$$\|v_1\| = \sqrt{(1)^2 + (1)^2 + (1)^2 + (0)^2 + (0)^2} = \sqrt{3}$$
$$\|v_2\| = \sqrt{(1)^2 + (0)^2 + (1)^2 + (1)^2 + (1)^2} = \sqrt{4} = 2$$
Do đó, độ tương đồng cosine là:
$$\text{Cosine}(v_1, v_2) = \frac{2}{\sqrt{3} \cdot 2} = \frac{1}{\sqrt{3}} \approx 0.577$$

> Joccard coefficient

Hệ số Jaccard được tính bằng công thức:
$$\text{Jaccard}(S_1, S_2) = \frac{|S_1 \cap S_2|}{|S_1 \cup S_2|}$$
Ta đã tính được $|S_1 \cap S_2| = 2$ và $|S_1 \cup S_2| = 5$, do đó:
$$\text{Jaccard}(S_1, S_2) = \frac{2}{5} = 0.4$$


###  5. Cho $\overline{X}$ và $\overline{Y}$ là hai điểm dữ liệu. Chứng minh rằng góc cosine giữa hai vector $\overline{X}$ và $\overline{Y}$ được tính bằng công thức 
$$\text{cosine}(\overline{X}, \overline{Y}) = \frac{||\overline{X}||^2 + ||\overline{Y}||^2 - ||\overline{X} - \overline{Y}||^2}{2||\overline{X}|| ||\overline{Y}||}$$

Xét 2 vector $\overline{X}$ và $\overline{Y}$ $\in \mathbb{R}^n$, ta có:
$$||\overline{X} - \overline{Y}||^2 = (\overline{X} - \overline{Y}) \cdot (\overline{X} - \overline{Y}) = \overline{X} \cdot \overline{X} - 2\overline{X} \cdot \overline{Y} + \overline{Y} \cdot \overline{Y}$$
Tiếp theo ta có:
$$||\overline{X}||^2 = \overline{X} \cdot \overline{X}$$
$$||\overline{Y}||^2 = \overline{Y} \cdot \overline{Y}$$
Thay vào công thức trên, ta có:
$$||\overline{X} - \overline{Y}||^2 = ||\overline{X}||^2 - 2\overline{X} \cdot \overline{Y} + ||\overline{Y}||^2$$
Sắp xếp lại công thức, ta có:
$$2\overline{X} \cdot \overline{Y} = ||\overline{X}||^2 + ||\overline{Y}||^2 - ||\overline{X} - \overline{Y}||^2$$
Do đó, ta có:
$$\overline{X} \cdot \overline{Y} = \frac{||\overline{X}||^2 + ||\overline{Y}||^2 - ||\overline{X} - \overline{Y}||^2}{2}$$
Mà góc cosine giữa hai vector $\overline{X}$ và $\overline{Y}$ được tính bằng công thức:
$$\text{cosine}(\overline{X}, \overline{Y}) = \frac{\overline{X} \cdot \overline{Y}}{||\overline{X}|| ||\overline{Y}||}$$
Thay vào công thức trên, ta có:
$$\text{cosine}(\overline{X}, \overline{Y}) = \frac{\frac{||\overline{X}||^2 + ||\overline{Y}||^2 - ||\overline{X} - \overline{Y}||^2}{2}}{||\overline{X}|| ||\overline{Y}||} = \frac{||\overline{X}||^2 + ||\overline{Y}||^2 - ||\overline{X} - \overline{Y}||^2}{2||\overline{X}|| ||\overline{Y}||}$$
