### Bài 1

- Công thức absolute support: $$\text{abs\_support}(X) = |\{t \in T \mid X \subseteq t\}|$$
    - Trong đó, $T$ là tập hợp tất cả các giao dịch
- Công thức relative support: $$\text{rel\_support}(X) = \frac{\text{abs\_support}(X)}{\text{N}}$$
    - Trong đó, $N$ là tổng số giao dịch trong tập dữ liệu.

Tổng số giao dịch N = 5
- $A = \{a, e, f\}$
    - Absolute support của $A = \{a, e, f\} = 2$ (giao dịch 3 và 4)
    - Relative support của $A = \{a, e, f\} = \frac{2}{5} = 0.4$
- $B = \{d, f\}$
    - Absolute support của $B = \{d, f\} = 2$ (giao dịch 3 và 5)
    - Relative support của $B = \{d, f\} = \frac{2}{5} = 0.4$

### Bài 7

Công thức tính confidence: $$\text{confidence}(X \rightarrow Y) = \frac{\text{abs\_support}(X \cup Y)}{\text{abs\_support}(X)}$$

Ta có:
- abs\_support($\{a\}$) = 3 (1, 3 và 4)
- abs\_support($\{a, f\}$) = 2 (3, 4)

confidence($\{a\} \rightarrow \{f\}$) = $\frac{2}{3} \approx 66.67\%$

- abs\_support($\{a, e\}$) = 2 (3, 4)
- abs\_support($\{a, e, f\}$) = 2 (3, 4)

confidence($\{a, e\} \rightarrow \{f\}$) = $\frac{2}{2} = 100\%$
