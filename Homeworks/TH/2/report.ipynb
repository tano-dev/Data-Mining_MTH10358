#### Viết hàm tính các chuẩn p = 1, 2, inf cho 50 dòng đầu tiên của array trong mục 1


````python
def calculate_distances(array):
    distances = []
    for i in range(50):
        for j in range(i + 1, 50):
            point_i = array[i, :]
            point_j = array[j, :]
            dist_p1 = np.linalg.norm(point_i - point_j, 1)
            dist_p2 = np.linalg.norm(point_i - point_j, 2)
            dist_pinf = np.linalg.norm(point_i - point_j, np.inf)
            distances.append((i, j, dist_p1, dist_p2, dist_pinf))
    return distances




distances = calculate_distances(array)
for dist in distances:
    print(f"Distance between point {dist[0]} and point {dist[1]}: p=1: {dist[2]}, p=2: {dist[3]}, p=inf: {dist[4]}")
    
````

Do yêu cầu đề bài tập là viết hàm tính các chuẩn p = 1, 2, inf cho 50 dòng đầu tiên của array. Nên trong hàm 'calculate_distances' ta sẽ chạy vòng lặp for để tính khoảng cách giữa array với 50 dòng đầu tiên của array đó. 

```python

    for j in range(i + 1, 50):
         ... # Tính khoảng cách giữa dòng i và dòng j
    ...
```

Note: j bắt đầu từ i + 1 để tránh tính khoảng cách giữa một dòng với chính nó và tránh tính lại khoảng cách đã tính trước đó (do khoảng cách giữa dòng i và dòng j là như nhau với khoảng cách giữa dòng j và dòng i).

Tiếp theo chúng ta trích xuất dữ liệu của điểm i và j để tạo thành 2 vector `point_i` và `point_j`:

```python

```
Sau đó, chúng ta sẽ tính khoảng cách giữa `point_i` và `point_j` theo các chuẩn p = 1, 2, inf... bằng cách sử dụng hàm `np.linalg.norm(vector, ord=p)` của NumPy:
Với p = 1 là khoảng cách Manhattan, p = 2 là khoảng cách Euclidean, và p = inf là khoảng cách Chebyshev. Cụ thể:

```python

dist_p2 = np.linalg.norm(point_i - point_j, 2)  # Khoảng cách Euclidean
dist_pinf = np.linalg.norm(point_i - point_j, np.inf)  # Khoảng cách Chebyshev
```

Sau khi tính toán xong thì ta lưu lại vào `distances` để sau này có thể sử dụng hoặc in ra kết quả.

```python

```

Cuối cùng ta được hàm `calculate_distances` như yêu cầu:
input: array (dữ liệu đã được xử lý)
output: distances (danh sách chứa khoảng cách giữa các điểm)

Tiếp theo chúng ta tạo ma trận n x n để lưu trữ độ đo tương đồng giữa các điểm, trong đó n là số dòng của dữ liệu (ở đây là 100 do đề bài). 
- `overlap_matrix` để lưu trữ số lượng thuộc tính giống nhau giữa các điểm.
- `inv_freq_matrix` để lưu trữ độ đo tương đồng dựa trên tần suất nghịch đảo của các giá trị thuộc tính.


````python
n = len(df_100)
overlap_matrix = np.zeros((n, n))
inv_freq_matrix = np.zeros((n, n))
````

Tiếp theo ta tính độ tương đồng giữa các điểm trong tập dữ liệu với công thức tổng quát như sau:

$$\text{Sim}(\overline{X}, \overline{Y} ) = \sum_{i=1}^{d} w_i \cdot \text{S}(x_i, y_i)$$
Với
- $w_i = \frac{1}{d}$ với i = 1, 2, ..., d
- $S(x_i, y_i)$ là hàm đánh giá sự giống nhau giữa thuộc tính $i$ được chia làm 2 loại là độ đo Overlap và độ đo dựa trên tần suất xuất hiện ngược


Ta tính độ đo Overlap với công thức: 

$$\text{Overlap}(x_i, y_i) = \text{S}(x_i, y_i) = \begin{cases} 1 & \text{nếu } x_i = y_i \\ 0 & \text{nếu } x_i \neq y_i \end{cases}$$


Cụ thể với code như sau:
`d = len(X)` để lấy số lượng thuộc tính của điểm X (và Y do |X| = |Y|).

Tiếp theo ta tính độ đo Overlap giữa X và Y bằng cách đếm số lượng thuộc tính giống nhau giữa hai điểm và chia cho tổng số thuộc tính d như công thức tổng quát sau khi thay $S(x_i, y_i)$ bằng công thức Overlap:
```python
sum = 0
for i in range(d):
    if X[i] == Y[i]:
        sum += 1
return sum / d
```


````python
def overlap(X, Y):
    d = len(X)
    sum = 0
    for i in range(d):
        if X[i] == Y[i]:
            sum += 1
    return sum / d
````

Tương tự ta tính độ đo dựa trên tần suất xuất hiện ngược với công thức:

$$\text{Invfreq}(x_i, y_i) = \text{S}(x_i, y_i) = \begin{cases} \frac{1}{1 + \text{log}(f_i(x_i)) \cdot \text{log}(f_i(y_i))} & \text{nếu } x_i = y_i \\ 0 & \text{nếu } x_i \neq y_i \end{cases}$$

Note: $log$ ở đây là log cơ số 10, và $f_i(x_i)$ là tần suất xuất hiện của giá trị $x_i$ trong thuộc tính $i$ của tập dữ liệu

Tạo một hàm riêng để đếm và lấy tần suất xuất hiện của các giá trị thuộc tính trong tập dữ liệu để có thể sử dụng trong hàm tính độ đo dựa trên tần suất xuất hiện ngược.
Bằng cách ta dùng kỹ thuật dictionary để lưu tên của thuộc tính và tần suất xuất hiện của các giá trị thuộc tính đó trong tập dữ liệu.

```python
    for i in range(d):
        unique, counts = np.unique(data[:, i], return_counts=True)
        freq_dict = {} 
        for j in range(len(unique)):
            keyname = unique[j]       
            value = counts[j] 
            freq_dict[keyname] = value        
        freqs.append(freq_dict)
    return freqs
```

`unique` là mảng chứa các giá trị duy nhất của thuộc tính i, `counts` là mảng chứa số lần xuất hiện của từng giá trị đó. Ta tạo một dictionary `freq_dict` để lưu trữ tần suất xuất hiện của từng giá trị, với key là giá trị và value là số lần xuất hiện. Sau đó ta thêm `freq_dict` vào list `freqs` để có thể sử dụng sau này trong hàm tính độ đo dựa trên tần suất xuất hiện ngược.

````python
def freq(data):
    data = data.values
    freqs = []
    d = data.shape[1]
    
    for i in range(d):
        unique, counts = np.unique(data[:, i], return_counts=True)
        freq_dict = {} 
        for j in range(len(unique)):
            keyname = unique[j]       
            value = counts[j] 
            freq_dict[keyname] = value        
        freqs.append(freq_dict)
    return freqs
````

Cuối cùng ta áp dụng công thức tính độ đo dựa trên tần suất xuất hiện ngược vào hàm `inv_freq` bằng cách thay $S(x_i, y_i)$ bằng công thức đã cho và sử dụng tần suất xuất hiện từ `freqs` 

````python
def inv_freq(X, Y, freqs):
    d = len(X)
    sim = 0
    
    for i in range(d):
        if X[i] == Y[i]:
            f_x = freqs[i].get(X[i], 0)
            f_y = freqs[i].get(Y[i], 0)
            
            if f_x > 0 and f_y > 0:
                sim += 1 / (1 + (np.log(f_x) * np.log(f_y)))
    return sim / d
````

````python
freqs = freq(df_100)

for i in range(n):
    for j in range(n):
        if i == j:
            overlap_matrix[i, j] = 1.0 
            inv_freq_matrix[i, j] = 1.0
        else:
            overlap_matrix[i, j] = overlap(df_100.values[i], df_100.values[j])
            inv_freq_matrix[i, j] = inv_freq(df_100.values[i], df_100.values[j], freqs)

for i in range(n): 
    overlap_scores = overlap_matrix[i].copy()
    inv_freq_scores = inv_freq_matrix[i].copy()
    
    overlap_scores[i] = -1 
    inv_freq_scores[i] = -1
    
    best_overlap_idx = np.argmax(overlap_scores)
    best_inv_freq_idx = np.argmax(inv_freq_scores)
    
    print(f"\n df line {i}:")
    print(f" - Overlap ~= {best_overlap_idx}, v = {overlap_scores[best_overlap_idx]}")
    print(f" - Inverse freq ~= {best_inv_freq_idx}, v = {inv_freq_scores[best_inv_freq_idx]}")
````
